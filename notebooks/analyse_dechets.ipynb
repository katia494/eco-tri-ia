## 1. Importation des bibliothèques

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from PIL import Image
import os
import json

print('Bibliothèques chargées avec succès ✅')
```

## 2. Distribution des classes de déchets

```python
# Classes de déchets et distribution simulée
classes = ['Carton', 'Verre', 'Métal', 'Papier', 'Plastique', 'Déchets généraux']
counts = [150, 120, 100, 180, 200, 90]

colors = ['#8B4513', '#2E8B57', '#708090', '#DEB887', '#FF6347', '#696969']

plt.figure(figsize=(10, 6))
bars = plt.bar(classes, counts, color=colors, edgecolor='black')
plt.title('Distribution des classes de déchets dans le dataset', fontsize=14)
plt.xlabel('Classe de déchet')
plt.ylabel('Nombre d\'images')
plt.xticks(rotation=15)

for bar, count in zip(bars, counts):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2,
             str(count), ha='center', fontsize=10)

plt.tight_layout()
plt.savefig('../data/processed/distribution_classes.png')
plt.show()
print(f'Total images: {sum(counts)}')
```

## 3. Métriques du modèle YOLOv8

```python
# Métriques de performance du modèle
metriques = {
    'Précision': 0.87,
    'Rappel': 0.83,
    'F1-Score': 0.85,
    'mAP@0.5': 0.89
}

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.barh(list(metriques.keys()), list(metriques.values()),
               color=['#2196F3', '#4CAF50', '#FF9800', '#9C27B0'])
ax.set_xlim(0, 1)
ax.set_title('Métriques de performance du modèle YOLOv8', fontsize=14)

for bar, val in zip(bars, metriques.values()):
    ax.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2,
            f'{val:.0%}', va='center', fontsize=11)

plt.tight_layout()
plt.savefig('../data/processed/metriques_modele.png')
plt.show()
print('Métriques affichées ✅')
```
